Numbers

1. Program to Find Odd or Even number

```python
x = int(input("Enter the number: "))

if x % 2 == 0:
    print("The given number is even.")
else:
    print("The given number is odd.")
```

2. Program to find Prime number

```python
x = int(input("Enter the number: "))

if x <= 1:
    print("The given number is not prime")
else:
    for i in range(2, x):
        if x % i == 0:
            print("The given number is not prime")
            break
    else:
        print("The given number is prime")
```

3. Program to find Fibonacci series upto a given number range

```python
x = int(input("Enter the range: "))

a = 0
b = 1

while a <= x:
    print(a)
    c = a + b
    a = b
    b = c
```

5. To find Factorial on given Number

```python
n = int(input("Enter the number: "))

a = 1

for i in range(1, n + 1):
    a = a * i

print(a)
```

6. To Reverse Number

```python
n = (input("Enter the number: "))

reverse = int(n[::-1])

print(reverse)
```

7. To find Armstrong Number

```python
n = int(input("Enter the number: "))

original = n
total = 0

while n > 0:
    a = n % 10
    total = total + a**3
    n = n // 10

if total == original:
    print("The given number is an Armstrong number")
else:
    print("The given number is not an Armstrong number")
```

8. To find number of digits in given number

```python
n = int(input("Enter the number: "))

count = 0

while n != 0:
    n = n // 10
    count = count + 1

print("Number of digits:", count)
```

9. To find Palindrome number

```python
n = int(input("Enter the number: "))

original = n
reverse = int(str(n)[::-1])

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

10. To calculate the sum of digits of a number

```python
a = input("Enter the number: ")

sum_of_numbers = 0

for i in a:
    sum_of_numbers = sum_of_numbers + int(i)

print("Sum of digits:", sum_of_numbers)
```
